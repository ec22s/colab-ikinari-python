```python
# @title (9) 検出した数を画像内に表示する準備（出力欄を画像上部に追加、本 6-4-2 同等）

# (7)と本 6-4-2（p.191）を元に自分でやってみる
# 画像上部に黒い出力欄が表示されればOK
# 出力欄の高さ extension_height が足りなければ適当に増やす
# (動画のサイズが大きくて再生時に縮小される時など)

# ↓ 完成例

from ultralytics import YOLO
import cv2
import numpy as np

def text_overwrite_to_image(img):
  """テキストを画像に描画する関数"""

  # 画像の横サイズを取得
  img_width = img.shape[1]

  # 黒い四角形（拡張領域）を作成
  extension_height = 40
  black_rectangle = np.zeros((extension_height, img_width, 3), dtype=np.uint8)

  # 元の画像と黒い四角形を結合
  extended_img = np.vstack((black_rectangle, img))

  return extended_img

model = YOLO("yolov8n.pt")
cap = cv2.VideoCapture("shibuya_live.mp4")

try:
  while cap.isOpened():
    # フレームを抽出
    ret, frame = cap.read()

    if ret:
      # 物体検出
      results = model(frame, verbose=False, classes=0)
      img_annotated = results[0].plot()

      # 分析（検出数をカウント）
      class_list = results[0].boxes.cls
      count_class = len(class_list)
      # まだ準備だけ

      # 画像にテキスト表示欄を追加
      img_annotated = text_overwrite_to_image(img_annotated)
      colab_imshow("png", img_annotated)

except KeyboardInterrupt:
  print("中止しました")
```

```python
# @title (10) 検出した数を画像内に表示（本 6-4-3 同等）

# (9)と本 6-4-3（p.192-193）を元に自分でやってみる
# 画像上部の出力欄に、検出した数が表示されればOK

# 出力欄の高さ extension_height が足りない場合、例えば下記のように関連のパラメータを変更
# (動画のサイズが大きくて再生時に縮小される時など)

# extension_height = 120
# font_scale = 2
# img_annotated = text_overwrite_to_image(..., (50, 75))

# 本来は出力欄の大きさに自動調整されるのが望ましい（時間あったら今後検討）

# ↓ 完成例

from ultralytics import YOLO
import cv2
import numpy as np

def text_overwrite_to_image(img, text, position):
  """テキストを画像に描画する関数"""

  # 画像の横サイズを取得
  img_width = img.shape[1]

  # 黒い四角形（拡張領域）を作成
  extension_height = 120
  black_rectangle = np.zeros((extension_height, img_width, 3), dtype=np.uint8)

  # 元の画像と黒い四角形を結合
  extended_img = np.vstack((black_rectangle, img))

  # 文字の設定
  font = cv2.FONT_HERSHEY_SIMPLEX
  font_scale = 2
  font_color = (255, 255, 255)
  font_thickness = 2

  # 画像にテキストを上書き
  cv2.putText(extended_img,
              text=text,
              org=position,
              fontFace=font,
              fontScale=font_scale,
              color=font_color,
              thickness=font_thickness)

  return extended_img

model = YOLO("yolov8n.pt")
cap = cv2.VideoCapture("shibuya_live.mp4")

try:
  while cap.isOpened():
    # フレームを抽出
    ret, frame = cap.read()

    if ret:
      # 物体検出
      results = model(frame, verbose=False, classes=0)
      img_annotated = results[0].plot()

      # 分析（検出数をカウント）
      class_list = results[0].boxes.cls
      count_class = len(class_list)

      # 画像に検出数を表示
      img_annotated = text_overwrite_to_image(img_annotated, f"person={count_class}", (50, 75))
      colab_imshow("png", img_annotated)

except KeyboardInterrupt:
  print("中止しました")
```

```python
# @title (11) 独自の応用例：検出したクラスと数をまとめて表示

# 本 6-4-3 で作った関数 text_overwrite_to_image の第2引数 text に渡す文字列を豊富化する
# 検出対象のクラスは限定せず、検出したクラスと数を全て表示する
# 例えば下記のような処理を物体検出後に追加し、text_overwrite_to_image に counts を渡す

# result = results[0]
# names_count = {}
# for box, cls in zip(result.boxes, result.boxes.cls):
#   name = result.names[int(cls)]
#   if name in names_count:
#     names_count[name] += 1
#   else:
#     names_count[name] = 1

# counts = ", ".join(
#   [f"{name}={names_count[name]}" for name in sorted(names_count.keys())]
# )

# (9)または(10)を元に自分でやってみる

# ↓ 完成例

from ultralytics import YOLO
import cv2
import numpy as np

def text_overwrite_to_image(img, text, position):
  """テキストを画像に描画する関数"""

  # 画像の横サイズを取得
  img_width = img.shape[1]

  # 黒い四角形（拡張領域）を作成
  extension_height = 120
  black_rectangle = np.zeros((extension_height, img_width, 3), dtype=np.uint8)

  # 元の画像と黒い四角形を結合
  extended_img = np.vstack((black_rectangle, img))

  # 文字の設定
  font = cv2.FONT_HERSHEY_SIMPLEX
  font_scale = 2
  font_color = (255, 255, 255)
  font_thickness = 2

  # 画像にテキストを上書き
  cv2.putText(extended_img,
              text=text,
              org=position,
              fontFace=font,
              fontScale=font_scale,
              color=font_color,
              thickness=font_thickness)

  return extended_img

model = YOLO("yolov8n.pt")
cap = cv2.VideoCapture("shibuya_live.mp4")

try:
  while cap.isOpened():
    # フレームを抽出
    ret, frame = cap.read()

    if ret:
      # 物体検出（対象クラスの限定なし）
      results = model(frame, verbose=False)
      img_annotated = results[0].plot()

      # 分析（対象クラス名ごとに検出数をカウント）
      result = results[0]
      names_count = {}
      for box, cls in zip(result.boxes, result.boxes.cls):
        name = result.names[int(cls)]
        if name in names_count:
          names_count[name] += 1
        else:
          names_count[name] = 1

      counts = ", ".join(
        [f"{name}={names_count[name]}" for name in sorted(names_count.keys())]
      )

      # 画像に検出数を表示
      img_annotated = text_overwrite_to_image(img_annotated, counts, (50, 75))
      colab_imshow("png", img_annotated)

except KeyboardInterrupt:
  print("中止しました")
```
