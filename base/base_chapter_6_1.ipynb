```python
# @title (5) 動画から物体検出する準備（学習会用の独自関数 colab_imshow を追加）

# 説明：
# 本と同様に動画からリアルタイムに物体検出し結果の画像を表示する準備として、独自関数を追加した
# 刻々変わる画像を切り替えて紙芝居のように表示する
# 前セルで使った ColabCap2 と組み合わせると、撮影中の動画（滑らかな表示）に加え、動画から抽出
# した各フレームがやや遅れて表示される
# ネット環境等によっては遅れが大きくなるかも
# この「フレーム抽出」〜「画像表示」の間に物体検出を挿入するのが次のセル

# 補足：
# - Googleによる関数 cv2_imshow は実質使えない（表示した画像が全て出力欄に残ってしまう）
# - 独自関数 colab_imshow は表示する画像のフォーマットを第1引数に指定する
#   jpegとpngはほぼ差がないが、gifにすると色数の少ないザラついた画像になる

# テスト：
# このセルを実行して動作を確認する
# 左にカメラ撮影中の動画（滑らかな表示）、右に抽出したフレームがやや遅れて表示されればOK
# colab_imshow の第1引数を gif に変えて違いを見る
# 終了は前セルと同様、セルの実行中止ボタンで

import requests

branch = "https://github.com/ec22s/colab-ikinari-python/raw/refs/heads/main"
paths = [ "util/ColabCap2.py", "util/colab_imshow.py" ]
for path in paths:
  exec(requests.get(f"{branch}/{path}", allow_redirects=True).content)

cap = ColabCap2()

try:
  while cap.isOpened:
    # フレームを抽出
    ret, frame = cap.read()

    if ret:
      # 画像を表示
      colab_imshow("png", frame)
except KeyboardInterrupt:
  print("ColabCap stopped.")

cap.release()
```
